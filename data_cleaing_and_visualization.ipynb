```python
# Convert text columns to string
df["ORDER_ID"] = df["ORDER_ID"].astype("string")
df["CUSTOMER_ID"] = df["CUSTOMER_ID"].astype("string")

# Convert Order_Date to datetime
df["ORDER_DATE"] = pd.to_datetime(df["ORDER_DATE"])

# Display the updated datatypes
print(df.dtypes)
```

```python
df["CITY"] = df["CITY"].astype("string")
df["RESTAURANT_NAME"] = df["RESTAURANT_NAME"].astype("string")
df["CUISINE_TYPE"] = df["CUISINE_TYPE"].astype("string")
df["ORDER_TYPE"] = df["ORDER_TYPE"].astype("string")
df["PAYMENT_METHOD"] = df["PAYMENT_METHOD"].astype("string")
df["ORDER_STATUS"] = df["ORDER_STATUS"].astype("string")
```

```python
# Missing payment methods
df["PAYMENT_METHOD"] = df["PAYMENT_METHOD"].fillna("Unknown")

# Separate delivery and pickup orders
delivery_mask = df["ORDER_TYPE"].eq("Delivery")
pickup_mask = df["ORDER_TYPE"].eq("Pickup")

# Missing delivery time for pickup orders -> 0
df.loc[pickup_mask, "DELIVERY_TIME_MINUTES"] = (
    df.loc[pickup_mask, "DELIVERY_TIME_MINUTES"].fillna(0)
)

# Fill missing delivery times using the median for each city
city_medians = (
    df.loc[delivery_mask]
      .groupby("CITY")["DELIVERY_TIME_MINUTES"]
      .transform("median")
)

df.loc[delivery_mask, "DELIVERY_TIME_MINUTES"] = (
    df.loc[delivery_mask, "DELIVERY_TIME_MINUTES"]
      .fillna(city_medians)
)

# Fallback if a city's median is unavailable
overall_median = df.loc[
    delivery_mask, "DELIVERY_TIME_MINUTES"
].median()

df.loc[delivery_mask, "DELIVERY_TIME_MINUTES"] = (
    df.loc[delivery_mask, "DELIVERY_TIME_MINUTES"]
      .fillna(overall_median)
)

print("Remaining missing values:")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Standardize column names
df.columns = df.columns.str.upper()

# Standardize text values used in filters
for col in ["ORDER_STATUS", "ORDER_TYPE"]:
    df[col] = df[col].astype("string").str.strip().str.upper()

# Convert date and time
df["ORDER_DATE"] = pd.to_datetime(
    df["ORDER_DATE"], errors="coerce"
)

df["ORDER_HOUR"] = pd.to_datetime(
    df["ORDER_TIME"],
    format="%H:%M",
    errors="coerce"
).dt.hour

sns.set_theme(style="whitegrid")
```

```python
DELIVERED = df[df["ORDER_STATUS"] == "DELIVERED"]

TOTAL_ORDERS = df["ORDER_ID"].nunique()
DELIVERED_COUNT = DELIVERED["ORDER_ID"].nunique()

TOTAL_REVENUE = DELIVERED["TOTAL_AMOUNT_INR"].sum()

AVERAGE_ORDER_VALUE = (
    DELIVERED["TOTAL_AMOUNT_INR"].mean()
)

CANCELLATION_RATE = (
    (df["ORDER_STATUS"] == "CANCELLED").mean() * 100
)

AVERAGE_RATING = DELIVERED["CUSTOMER_RATING"].mean()

print("ZOMATO ORDERS — KPI SUMMARY")
print("-" * 35)
print(f"Total Orders: {TOTAL_ORDERS:,}")
print(f"Delivered Orders: {DELIVERED_COUNT:,}")
print(f"Delivered-Order Revenue: ₹{TOTAL_REVENUE:,.2f}")
print(f"Average Order Value: ₹{AVERAGE_ORDER_VALUE:,.2f}")
print(f"Cancellation Rate: {CANCELLATION_RATE:.2f}%")
print(f"Average Customer Rating: {AVERAGE_RATING:.2f}/5")
```

```python
WEEKDAY_ORDER = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

df["WEEKDAY"] = df["ORDER_DATE"].dt.day_name()

HEATMAP_DATA = pd.crosstab(
    df["WEEKDAY"],
    df["ORDER_HOUR"]
).reindex(WEEKDAY_ORDER)

plt.figure(figsize=(14, 6))

sns.heatmap(
    HEATMAP_DATA,
    cmap="YlOrRd",
    linewidths=0.3
)

plt.title("Zomato Orders by Weekday and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Weekday")
plt.tight_layout()
plt.show()
```

```python
CITY_STATUS = pd.crosstab(
    df["CITY"],
    df["ORDER_STATUS"],
    normalize="index"
) * 100

CANCEL_RATES = (
    CITY_STATUS["CANCELLED"]
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=CANCEL_RATES.index,
    y=CANCEL_RATES.values
)

plt.title("Cancellation Rate by City")
plt.xlabel("City")
plt.ylabel("Cancelled Orders (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
CUSTOMER_STATS = (
    DELIVERED.groupby("CUSTOMER_ID")
    .agg(
        ORDER_COUNT=("ORDER_ID", "nunique"),
        TOTAL_SPEND=("TOTAL_AMOUNT_INR", "sum"),
        AVG_ORDER_VALUE=("TOTAL_AMOUNT_INR", "mean")
    )
    .sort_values("TOTAL_SPEND", ascending=False)
)

print("Top 10 customers by spending:")
display(CUSTOMER_STATS.head(10))

REPEAT_CUSTOMERS = (
    CUSTOMER_STATS["ORDER_COUNT"] > 1
).sum()

TOTAL_CUSTOMERS = len(CUSTOMER_STATS)

REPEAT_RATE = (
    REPEAT_CUSTOMERS / TOTAL_CUSTOMERS * 100
    if TOTAL_CUSTOMERS else 0
)

print(f"Unique delivered-order customers: {TOTAL_CUSTOMERS}")
print(f"Customers with repeat orders: {REPEAT_CUSTOMERS}")
print(f"Repeat-customer percentage: {REPEAT_RATE:.2f}%")
```

```python
TOP_CUSTOMERS = CUSTOMER_STATS.head(10).reset_index()

plt.figure(figsize=(11, 5))

sns.barplot(
    data=TOP_CUSTOMERS,
    x="CUSTOMER_ID",
    y="TOTAL_SPEND"
)

plt.title("Top 10 Customers by Total Spending")
plt.xlabel("Customer ID")
plt.ylabel("Total Spending (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
RATING_DATA = df[
    (df["ORDER_STATUS"] == "DELIVERED")
    & (df["ORDER_TYPE"] == "DELIVERY")
    & (df["CUSTOMER_RATING"].notna())
    & (df["DELIVERY_TIME_MINUTES"].notna())
]

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=RATING_DATA,
    x="CUSTOMER_RATING",
    y="DELIVERY_TIME_MINUTES"
)

plt.title("Delivery Time by Customer Rating")
plt.xlabel("Customer Rating")
plt.ylabel("Delivery Time (Minutes)")
plt.tight_layout()
plt.show()
```

```python
DISCOUNT_ANALYSIS = (
    DELIVERED.groupby("DISCOUNT_PERCENT")
    .agg(
        ORDER_COUNT=("ORDER_ID", "nunique"),
        AVG_ORDER_VALUE=("TOTAL_AMOUNT_INR", "mean")
    )
    .reset_index()
)

fig, ax1 = plt.subplots(figsize=(10, 5))

sns.barplot(
    data=DISCOUNT_ANALYSIS,
    x="DISCOUNT_PERCENT",
    y="AVG_ORDER_VALUE",
    ax=ax1
)

ax1.set_title("Average Order Value by Discount Percentage")
ax1.set_xlabel("Discount (%)")
ax1.set_ylabel("Average Order Value (INR)")

plt.tight_layout()
plt.show()

display(DISCOUNT_ANALYSIS)
```
